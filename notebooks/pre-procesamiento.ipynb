# **Limpieza de datos**

```python
import pandas as pd

df = pd.read_csv('../data/processed/listings_subset.csv')
```

```python
# Convertir price de str a float

# Paso 1: Sacar $ y espacios
df['price'] = df['price'].str.replace('$', '', regex=False)
df['price'] = df['price'].str.replace(',', '', regex=False)
df['price'] = df['price'].str.strip()  # Sacar espacios

# Paso 2: Convertir a número (float)
df['price'] = pd.to_numeric(df['price'], errors='coerce')
```

```python
# Ver cantidad de NA
nulos = df.isnull().sum()
print(nulos)
```

```python
# Eliminar filas que tengan NA en estos atributos importantes
df = df.dropna(subset=['price','neighbourhood_cleansed','room_type'])

# Fill NA con 0
df['number_of_reviews'] = df['number_of_reviews'].fillna(0)
df['availability_365'] = df['availability_365'].fillna(0)

# Si no hay rating, se deja en 0
df['review_scores_rating'] = df['review_scores_rating'].fillna(0)

# Si no hay mínimo de noches, se deja en 1
df['minimum_nights'] = df['minimum_nights'].fillna(1)

# Provisorio
df['bedrooms'] = df['bedrooms'].fillna(1)
df['beds'] = df['beds'].fillna(1)
df['bathrooms'] = df['bathrooms'].fillna(1)
df['description'] = df['description'].fillna('')
df['last_review'] = pd.to_datetime(df['last_review'])

# Ver cantidad de NA
nulos = df.isnull().sum()
print(nulos)
```
